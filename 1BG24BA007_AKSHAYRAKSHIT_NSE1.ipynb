```python
contacts = [
    {"id": 1, "name": "Akshay", "phone": "9876543210", "city": "Bangalore"},
    {"id": 2, "name": "Ravi", "phone": "9123456780", "city": "Mysore"},
    {"id": 3, "name": "Ananya", "phone": "9988776655", "city": "Chennai"},
    {"id": 4, "name": "Rahul", "phone": "9012345678", "city": "Delhi"},
    {"id": 5, "name": "Priya", "phone": "9090909090", "city": "Mumbai"},
    {"id": 6, "name": "Suresh", "phone": "8888777766", "city": "Hyderabad"},
    {"id": 7, "name": "Neha", "phone": "7777666655", "city": "Pune"},
    {"id": 8, "name": "Arjun", "phone": "6666555544", "city": "Kochi"},
    {"id": 9, "name": "Kavya", "phone": "9555443322", "city": "Udupi"},
    {"id": 10, "name": "Manoj", "phone": "9444332211", "city": "Hubli"}
]

contacts
```

```python
# View all contacts
def view_contacts():
    if len(contacts) == 0:
        print("No contacts available.")
    else:
        print("\n--- Contact List ---")
        for c in contacts:
            print("ID:", c["id"],
                  "| Name:", c["name"],
                  "| Phone:", c["phone"],
                  "| City:", c["city"])
```

```python

# Search contact by name
def search_contact():
    name = input("Enter name to search: ").strip().lower()
    found = False

    for c in contacts:
        if c["name"].lower() == name:
            print("\nContact Found:")
            for k, v in c.items():
                print(k, ":", v)
            found = True
            break

    if not found:
        print("Contact not found.")
```

```python
# Add new contact
def add_contact():
    try:
        cid = int(input("Enter Contact ID: "))
        name = input("Enter Name: ").strip()
        phone = input("Enter Phone: ").strip()
        city = input("Enter City: ").strip()

        new_contact = {
            "id": cid,
            "name": name,
            "phone": phone,
            "city": city
        }

        contacts.append(new_contact)
        print("Contact added successfully.")

    except ValueError:
        print("Invalid ID! Please enter a number.")
```

```python
# Save contacts to file
def save_to_file():
    with open("contacts.txt", "w") as file:
        for c in contacts:
            file.write(f"{c['id']},{c['name']},{c['phone']},{c['city']}\n")

    if os.path.exists("contacts.txt"):
        print("Contacts saved to file successfully.")
while True:
    print("\n--- Contact Book Menu ---")
    print("1. View All Contacts")
    print("2. Search Contact")
    print("3. Add Contact")
    print("4. Save Contacts to File")
    print("5. Exit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        view_contacts()
    elif choice == "2":
        search_contact()
    elif choice == "3":
        add_contact()
    elif choice == "4":
        save_to_file()
    elif choice == "5":
        print("Thank you! Exiting Contact Book.")
        break
    else:
        print("Invalid choice. Please try again.")
```
